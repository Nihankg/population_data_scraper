```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
```

```python
df = pd.read_excel(
    "/Users/nihankaragul/Documents/Documents - Nihan’s MacBook Air/2024 Fall/Dissertation/Data/Processed/rdd_data_2.xlsx"
)
```

```python
collapse = gpd.read_file(
    "/Users/nihankaragul/Documents/Documents - Nihan’s MacBook Air/2024 Fall/Dissertation/Data/collapse_desc/collapse_desc.shp"
)
```

```python
collapse_merge = collapse.merge(
    df,
    left_on=["Provinc", "Distrct"],
    right_on=["Provinc", "Distrct"],
    how="inner"  # or 'left', depending on what you want
)
```

```python
borders = gpd.read_file(
    "/Users/nihankaragul/Documents/Documents - Nihan’s MacBook Air/2024 Fall/Dissertation/Data/borders/borders.shp"
)
```

```python
collapse_merge
```

```python

hatay = collapse_merge[collapse_merge["Provinc"] == "hatay"].copy()

# --- 1. Define district names ---
hatay_districts = [
    "antakya", "defne", "samandag", "iskenderun", "altinozu", "hassa", 
    "dortyol", "erzin", "kırikhan", "kumlu", "payas", "reyhanli", "belen", "yayladagi", "arsuz"
]

highlight_districts = ["altinozu", "arsuz", "defne", "iskenderun", "payas", "reyhanli"]

# --- 2. Prepare data ---
# Make sure district names are lowercase for matching
borders["name"] = borders["name"].str.lower()

# Filter only Hatay districts
borders_hatay = borders[borders["name"].isin(hatay_districts)]
highlight_layer = borders[borders["name"].isin(highlight_districts)]

# Compute log(total damage)
hatay["log_total_dmg"] = np.log1p(hatay["ttl_bld_x"])  # log(x+1) avoids issues with zeros

# --- 3. Define custom red color gradient ---
custom_reds = LinearSegmentedColormap.from_list(
    "custom_reds",
    [
        (1, 0.95, 0.95),  # very light pink
        (1, 0.7, 0.7),    # soft red
        (0.9, 0.3, 0.3),  # medium red
        (0.5, 0, 0)       # dark red
    ]
)

# --- 4. Plot ---
fig, ax = plt.subplots(figsize=(12, 12))

# Base layer — continuous red shades for log(total damage)
hatay.plot(
    ax=ax,
    column="log_total_dmg",
    cmap=custom_reds,
    legend=True,
    edgecolor="gray",
    linewidth=0.2,
    legend_kwds={
        "label": "Log(Total Damage + 1)",
        "shrink": 0.6,  # control size of colorbar
    },
    missing_kwds={
        "color": "white",
        "edgecolor": "lightgray",
        "label": "No data"
    }
)

# Overlay selected districts in semi-transparent green
highlight_layer.plot(ax=ax, color="limegreen", alpha=0.3)

# Overlay all Hatay district borders
borders_hatay.boundary.plot(ax=ax, color="black", linewidth=1.2)

# --- 5. Final layout and formatting ---
ax.set_title("Neighborhood-Level Log(Damage) — Hatay", fontsize=16)
ax.set_axis_off()

# Format colorbar text
cbar_ax = fig.axes[-1]
cbar_ax.tick_params(labelsize=10)
cbar_ax.set_ylabel("Log(Total Damage + 1)", fontsize=12)

# --- 6. Save and show ---
plt.savefig(
    "/Users/nihankaragul/Documents/Documents - Nihan’s MacBook Air/2024 Fall/Dissertation/Materials/collapse_map_hatay_log.pdf",
    format="pdf",
    bbox_inches="tight"
)
plt.show()
```

```python
# --- 1. Prepare data ---
# Compute log(total damage) for the entire dataset
collapse_merge["log_total_dmg"] = np.log1p(collapse_merge["ttl_bld_x"])  # log(x+1) avoids issues with zeros

# --- 2. Define custom red color gradient ---
custom_reds = LinearSegmentedColormap.from_list(
    "custom_reds",
    [
        (1, 0.95, 0.95),  # very light pink
        (1, 0.7, 0.7),    # soft red
        (0.9, 0.3, 0.3),  # medium red
        (0.5, 0, 0)       # dark red
    ]
)

# --- 3. Plot ---
fig, ax = plt.subplots(figsize=(14, 14))

# Plot all provinces using log damage
collapse_merge.plot(
    ax=ax,
    column="log_total_dmg",
    cmap=custom_reds,
    legend=True,
    edgecolor="gray",
    linewidth=0.2,
    legend_kwds={
        "label": "Log(Total Damage + 1)",
        "shrink": 0.6,
    },
    missing_kwds={
        "color": "white",
        "edgecolor": "lightgray",
        "label": "No data"
    }
)

# Overlay province borders (optional)
borders.boundary.plot(ax=ax, color="black", linewidth=0.6)

# --- 4. Layout and formatting ---
ax.set_title("Neighborhood-Level Log(Damage) — Entire Region", fontsize=18)
ax.set_axis_off()

# Format colorbar text
cbar_ax = fig.axes[-1]
cbar_ax.tick_params(labelsize=10)
cbar_ax.set_ylabel("Log(Total Damage + 1)", fontsize=12)

# --- 5. Save and show ---
plt.savefig(
    "/Users/nihankaragul/Documents/Documents - Nihan’s MacBook Air/2024 Fall/Dissertation/Materials/collapse_map_all_log.pdf",
    format="pdf",
    bbox_inches="tight"
)
plt.show()
```
